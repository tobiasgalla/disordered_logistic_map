```python
#importing data for theory lines -- produced in separate mathematica file
import numpy as np
import matplotlib.pyplot as plt



xdat = np.genfromtxt('x.csv', dtype = float, delimiter=',', skip_footer=0)

histvecth = np.genfromtxt('histvecth.csv', dtype = float, delimiter=',', skip_footer=0)
xvecth = np.genfromtxt('xvecth.csv', dtype = float, delimiter=',', skip_footer=0)

xmin = 0
xmax = 1.1
numbins = 51
binedges = np.linspace(xmin, xmax, numbins+1, endpoint=True)
bins = np.linspace(xmin+float((xmax-xmin)/(numbins-1))*0.5, xmax-float((xmax-xmin)/(numbins-1))*0.5, numbins, endpoint=True)		
xhist = np.histogram(xdat, binedges)[0]

n = 2000
numtrial = 20

#Plotting the eigenvalues and the theory lines
SIZE = 15

plt.rc('font', size=SIZE)          # controls default text sizes
plt.rc('axes', titlesize=SIZE)     # fontsize of the axes title
plt.rc('axes', labelsize=SIZE)    # fontsize of the x and y labels
plt.rc('xtick', labelsize=SIZE)    # fontsize of the tick labels
plt.rc('ytick', labelsize=SIZE)    # fontsize of the tick labels
plt.rc('legend', fontsize=SIZE-5)    # legend fontsize
plt.rc('text', usetex=True)
plt.rcParams['axes.linewidth'] = 1.25
plt.rcParams['legend.loc'] = 'best'

edgewidth = 1.5
lwidth = 1.25
size = 7.5
tickwidth = 1.25
ticklength = 5
padness = 5
fsize = 5

fig1 = plt.figure(figsize=(7.5,5))
ax1 = fig1.add_subplot(111)
ax1.plot(bins, xhist/float(n*numtrial)*float(numbins)/(xmax-xmin), 'x', markersize = (size -4.5), markeredgewidth = edgewidth)
ax1.plot(xvecth, histvecth, '-', color = 'r', markersize = (size -4.5), markeredgewidth = edgewidth)
axes = plt.gca()
axes.set_xlim([0.0,1.1])
axes.set_ylim([0.0, 1.5])
#axes.set_ylim([0,max(hist[0]/float(numtrial))])
ax1.set_xlabel(r"$x$")
ax1.set_ylabel(r"$P(x)$")
#ax1.legend()
ax1.tick_params(width = tickwidth, length = ticklength)
ax1.tick_params(pad=padness)
#ax1.text(0.8, 0.15, r"(a)", transform=ax1.transAxes, fontsize=fsize,
      		#verticalalignment='top', bbox=props, color='black')
fig1.show()

plt.savefig("histogram_points.svg", format="svg")

fig1 = plt.figure(figsize=(7.5,5))
ax1 = fig1.add_subplot(111)
ax1.bar(bins, xhist/float(n*numtrial)*float(numbins)/(xmax-xmin), width = (xmax-xmin)/float(numbins))
ax1.plot(xvecth, histvecth, '-', color = 'r', markersize = (size -4.5), markeredgewidth = edgewidth)
axes = plt.gca()
axes.set_xlim([0.0,1.1])
axes.set_ylim([0.0, 1.5])
#axes.set_ylim([0,max(hist[0]/float(numtrial))])
ax1.set_xlabel(r"$x$")
ax1.set_ylabel(r"$P(x)$")
#ax1.legend()
ax1.tick_params(width = tickwidth, length = ticklength)
ax1.tick_params(pad=padness)
#ax1.text(0.8, 0.15, r"(a)", transform=ax1.transAxes, fontsize=fsize,
      		#verticalalignment='top', bbox=props, color='black')
fig1.show()

plt.savefig("histogram_bars.svg", format="svg")
```
